```
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
import torchvision

to_tensor = torchvision.transforms.transforms.ToTensor()
train_dataset = torchvision.datasets.CIFAR10(root='./data/', train=True, download=True, transform=to_tensor)
test_dataset = torchvision.datasets.CIFAR10(root='./data/', train=False, download=True, transform=to_tensor)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
import matplotlib.pyplot as plt
import numpy as np

tensor_img = train_dataset[2][0].squeeze()
img = (tensor_img.permute(1, 2, 0).numpy() * 255.).astype(np.uint8)
plt.imshow(img)


def train(loader, model, criterion, optimizer):
    loss_per_epoch = []
    acc_per_epoch = []
    for data, label in loader:
        optimizer.zero_grad()
        data = data.view(-1, 3 * 32 * 32)  # Flattening image
        pred = model(data)
        loss = criterion(pred, label)
        loss.backward()
        optimizer.step()

        acc = accuracy_score(label, pred.argmax(dim=1))
        acc_per_epoch.append(acc)
        loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


def test(loader, model, criterion):
    loss_per_epoch = []
    acc_per_epoch = []
    with torch.no_grad():
        for data, label in loader:
            data = data.view(-1, 3 * 32 * 32)  # Flattening image
            pred = model(data)
            loss = criterion(pred, label)
            acc = accuracy_score(label, pred.argmax(dim=1))
            acc_per_epoch.append(acc)
            loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


3 * 32 * 32


class classifier(nn.Module):
    def __init__(self):
        super(classifier, self).__init__()
        self.fc1 = nn.Linear(3072, 10)

    def forward(self, x):
        x = self.fc1(x)
        return x


model = classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")
model = classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), momentum=0.1)  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")
https: // pytorch.org / docs / stable / generated / torch.optim.Adagrad.html
model = classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adagrad(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")
https: // pytorch.org / docs / stable / generated / torch.optim.RMSprop.html
model = classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.RMSprop(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")
https: // pytorch.org / docs / stable / generated / torch.optim.Adam.html
model = classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_Aacc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")
# 2025-03-02
## GPU Acceleration
import torch
import torch.nn as nn

torch.cuda.is_available()
a = torch.tensor(1.0)
a
b = torch.tensor(1.0)
b = b.to('cuda')
b
device = 'cuda'
a.to(device)
## Run model on GPU
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score
import torchvision
import numpy as np

to_tensor = torchvision.transforms.transforms.ToTensor()
train_dataset = torchvision.datasets.CIFAR10(root='./data/', train=True, download=True, transform=to_tensor)
test_dataset = torchvision.datasets.CIFAR10(root='./data/', train=False, download=True, transform=to_tensor)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)


def train(loader, model, criterion, optimizer, device):
    loss_per_epoch = []
    acc_per_epoch = []
    for data, label in loader:
        optimizer.zero_grad()
        data = data.view(-1, 3 * 32 * 32)  # Flattening image

        data = data.to(device)
        label = label.to(device)

        pred = model(data)
        loss = criterion(pred, label)
        loss.backward()
        optimizer.step()

        acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
        acc_per_epoch.append(acc)
        loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


def test(loader, model, criterion, device):
    loss_per_epoch = []
    acc_per_epoch = []
    with torch.no_grad():
        for data, label in loader:
            data = data.view(-1, 3 * 32 * 32)  # Flattening image

            data = data.to(device)
            label = label.to(device)

            pred = model(data)
            loss = criterion(pred, label)
            acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
            acc_per_epoch.append(acc)
            loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


class classifier(nn.Module):
    def __init__(self):
        super(classifier, self).__init__()
        self.fc1 = nn.Linear(3072, 10)

    def forward(self, x):
        x = self.fc1(x)
        return x


model = classifier()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters(), momentum=0.1)  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

model(torch.rand((1, 3072)))
model(torch.rand((1, 3072)).to(device))


class classifier(nn.Module):
    def __init__(self):
        super(classifier, self).__init__()
        self.fc1 = nn.Linear(32 * 32 * 3, 100)
        self.fc2 = nn.Linear(100, 10)
        self.relu1 = nn.ReLU()

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu1(x)
        x = self.fc2(x)
        return x


model = classifier()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters(), momentum=0.1)  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

model = classifier()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.Adam(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")


class classifier(nn.Module):
    def __init__(self):
        super(classifier, self).__init__()
        self.fc1 = nn.Linear(32 * 32 * 3, 500)
        self.fc2 = nn.Linear(500, 100)
        self.fc3 = nn.Linear(100, 10)
        self.relu1 = nn.ReLU()
        self.relu2 = nn.ReLU()

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu1(x)
        x = self.fc2(x)
        x = self.relu2(x)
        x = self.fc3(x)
        return x


32 * 32 * 3 * 500 + 500 * 100 + 100 * 10
model = classifier()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.Adam(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

import matplotlib.pyplot as plt

plt.plot(train_loss_per_epoch)
plt.plot(test_loss_per_epoch)
plt.plot(train_acc_per_epoch)
plt.plot(test_acc_per_epoch)
model = classifier()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

plt.plot(train_loss_per_epoch)
plt.plot(test_loss_per_epoch)
plt.plot(train_acc_per_epoch)
plt.plot(test_acc_per_epoch)
# Fashion MNIST
# Chinese MNIST
# CIFAR-100
# 2025-04-09
## Convolutional Neural Network
import torch
import torch.nn as nn
import torchvision
from torchvision import transforms
import numpy as np
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from sklearn.metrics import accuracy_score

to_tensor = torchvision.transforms.transforms.ToTensor()
train_dataset = torchvision.datasets.CIFAR10(root='./data/', train=True, download=True, transform=to_tensor)
test_dataset = torchvision.datasets.CIFAR10(root='./data/', train=False, download=True, transform=to_tensor)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)


def train(loader, model, criterion, optimizer, device):
    loss_per_epoch = []
    acc_per_epoch = []
    for data, label in loader:
        optimizer.zero_grad()

        data = data.to(device)
        label = label.to(device)

        pred = model(data)
        loss = criterion(pred, label)
        loss.backward()
        optimizer.step()

        acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
        acc_per_epoch.append(acc)
        loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


def test(loader, model, criterion, device):
    loss_per_epoch = []
    acc_per_epoch = []
    with torch.no_grad():
        for data, label in loader:
            data = data.to(device)
            label = label.to(device)

            pred = model(data)
            loss = criterion(pred, label)
            acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
            acc_per_epoch.append(acc)
            loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


4 * 4 * 32, 32 * 32 * 3


class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=8, kernel_size=3, stride=1, padding=1)  # 32*32*3 --> 32*32*8
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)  # 32*32*8 --> 16*16*8

        self.conv2 = nn.Conv2d(in_channels=8, out_channels=16, kernel_size=3, stride=1,
                               padding=1)  # 16*16*8 --> 16*16*16
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)  # 16*16*16 --> 8*8*16

        self.conv3 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, stride=1, padding=1)  # 8*8*16 --> 8*8*32
        self.relu3 = nn.ReLU()
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)  # 8*8*32 --> 4*4*32

        self.fc1 = nn.Linear(4 * 4 * 32, 10)  #

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)

        x = self.conv3(x)
        x = self.relu3(x)
        x = self.pool3(x)

        x = x.view(-1, 4 * 4 * 32)  # Flattening the Image (4,4,32)
        x = self.fc1(x)

        return x


model = CNN()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(20):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

import matplotlib.pyplot as plt

plt.plot(train_acc_per_epoch)
plt.plot(test_acc_per_epoch)


class CNNv2(nn.Module):
    def __init__(self):
        super(CNNv2, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, stride=1,
                               padding=1)  # 32*32*3 --> 32*32*16
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)  # 32*32*16 --> 16*16*16

        self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, stride=1,
                               padding=1)  # 16*16*16 --> 16*16*32
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)  # 16*16*32 --> 8*8*32

        self.conv3 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1, padding=1)  # 8*8*32 --> 8*8*64
        self.relu3 = nn.ReLU()
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)  # 8*8*64 --> 4*4*64

        self.fc1 = nn.Linear(4 * 4 * 64, 10)  #

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)

        x = self.conv3(x)
        x = self.relu3(x)
        x = self.pool3(x)

        x = x.view(-1, 4 * 4 * 64)  # Flattening the Image (4,4,32)
        x = self.fc1(x)

        return x


model = CNNv2()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(20):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

import matplotlib.pyplot as plt

plt.plot(train_acc_per_epoch)
plt.plot(test_acc_per_epoch)


class CNNv3(nn.Module):
    def __init__(self):
        super(CNNv3, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, stride=1,
                               padding=1)  # 32*32*3 --> 32*32*32
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)  # 32*32*32 --> 16*16*32

        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1,
                               padding=1)  # 16*16*32 --> 16*16*64
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)  # 16*16*64 --> 8*8*64

        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, stride=1,
                               padding=1)  # 8*8*64 --> 8*8*128
        self.relu3 = nn.ReLU()
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)  # 8*8*128 --> 4*4*128

        self.fc1 = nn.Linear(4 * 4 * 128, 10)  #

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)

        x = self.conv3(x)
        x = self.relu3(x)
        x = self.pool3(x)

        x = x.view(-1, 4 * 4 * 128)  # Flattening the Image (4,4,128)
        x = self.fc1(x)

        return x


model = CNNv3()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(20):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

import matplotlib.pyplot as plt

plt.plot(train_acc_per_epoch)
plt.plot(test_acc_per_epoch)


class CNNv3_1(nn.Module):
    def __init__(self):
        super(CNNv3_1, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, stride=1,
                               padding=1)  # 32*32*3 --> 32*32*32
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)  # 32*32*32 --> 16*16*32

        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1,
                               padding=1)  # 16*16*32 --> 16*16*64
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)  # 16*16*64 --> 8*8*64

        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, stride=1,
                               padding=1)  # 8*8*64 --> 8*8*128
        self.relu3 = nn.ReLU()
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)  # 8*8*128 --> 4*4*128

        self.fc1 = nn.Linear(4 * 4 * 128, 512)  #
        self.relu4 = nn.ReLU()
        self.fc2 = nn.Linear(512, 10)  #

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)

        x = self.conv3(x)
        x = self.relu3(x)
        x = self.pool3(x)

        x = x.view(-1, 4 * 4 * 128)  # Flattening the Image (4,4,32)
        x = self.fc1(x)
        x = self.relu4(x)
        x = self.fc2(x)

        return x


model = CNNv3_1()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(20):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

import matplotlib.pyplot as plt

plt.plot(train_acc_per_epoch)
plt.plot(test_acc_per_epoch)
# 2025-04-13
import torch
import torch.nn as nn
import torchvision
from torchvision import transforms
import numpy as np
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from sklearn.metrics import accuracy_score


class CNNv4(nn.Module):
    def __init__(self):
        super(CNNv4, self).__init__()
        self.block1 = self._make_block(3, 32)
        self.block2 = self._make_block(32, 64)
        self.block3 = self._make_block(64, 128)
        self.block4 = self._make_block(128, 256)

        self.fc1 = nn.Linear(2 * 2 * 256, 10)  #

    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)

        x = x.view(-1, 2 * 2 * 256)  # Flattening the Image (2,2,256)
        x = self.fc1(x)

        return x

    def _make_block(self, in_channels, out_channels):
        block = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        return block


def train(loader, model, criterion, optimizer, device):
    loss_per_epoch = []
    acc_per_epoch = []
    for data, label in loader:
        optimizer.zero_grad()

        data = data.to(device)
        label = label.to(device)

        pred = model(data)
        loss = criterion(pred, label)
        loss.backward()
        optimizer.step()

        acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
        acc_per_epoch.append(acc)
        loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


def test(loader, model, criterion, device):
    loss_per_epoch = []
    acc_per_epoch = []
    with torch.no_grad():
        for data, label in loader:
            data = data.to(device)
            label = label.to(device)

            pred = model(data)
            loss = criterion(pred, label)
            acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
            acc_per_epoch.append(acc)
            loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


to_tensor = torchvision.transforms.transforms.ToTensor()
train_dataset = torchvision.datasets.CIFAR10(root='./data/', train=True, download=True, transform=to_tensor)
test_dataset = torchvision.datasets.CIFAR10(root='./data/', train=False, download=True, transform=to_tensor)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
model = CNNv4()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(20):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

# 2025-05-16
import torch
import torch.nn as nn
import torchvision
from torchvision import transforms
import numpy as np
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from sklearn.metrics import accuracy_score


class CNNv4_1(nn.Module):
    def __init__(self):
        super(CNNv4_1, self).__init__()
        self.block1 = self._make_block(3, 32)
        self.block2 = self._make_block(32, 64)
        self.block3 = self._make_block(64, 128)
        # self.drop = nn.Dropout(p=0.5)
        self.fc1 = nn.Linear(4 * 4 * 128, 10)  #

    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)

        x = x.view(-1, 4 * 4 * 128)  # Flattening the Image (2,2,256)
        x = self.fc1(x)
        # x = self.relu(x)
        # x = self.d1(x)
        # x = self.fc2(x)
        # x = self.relu(x)
        # x = self.d2(x)
        # x = self.fc3(x)
        # x = self.relu(x)
        # x = self.d3(x)
        # x = self.fc4(x)

        return x

    def _make_block(self, in_channels, out_channels):
        block = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        return block


def train(loader, model, criterion, optimizer, device):
    loss_per_epoch = []
    acc_per_epoch = []
    for data, label in loader:
        optimizer.zero_grad()

        data = data.to(device)
        label = label.to(device)

        pred = model(data)
        loss = criterion(pred, label)
        loss.backward()
        optimizer.step()

        acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
        acc_per_epoch.append(acc)
        loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


def test(loader, model, criterion, device):
    loss_per_epoch = []
    acc_per_epoch = []
    with torch.no_grad():
        for data, label in loader:
            data = data.to(device)
            label = label.to(device)

            pred = model(data)
            loss = criterion(pred, label)
            acc = accuracy_score(label.detach().cpu(), pred.argmax(dim=1).detach().cpu())
            acc_per_epoch.append(acc)
            loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


to_tensor = torchvision.transforms.transforms.ToTensor()
train_dataset = torchvision.datasets.CIFAR10(root='./data/', train=True, download=True, transform=to_tensor)
test_dataset = torchvision.datasets.CIFAR10(root='./data/', train=False, download=True, transform=to_tensor)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
model = CNNv4_1()
criterion = nn.CrossEntropyLoss()

device = 'cuda'
model = model.to(device)
criterion = criterion.to(device)

optimizer = optim.SGD(model.parameters(), weight_decay=1e-6)  # w_new = w_old - lr * DL(w_old)

train_loss_per_epoch = []
train_acc_per_epoch = []
test_loss_per_epoch = []
test_acc_per_epoch = []
for epoch in range(20):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer, device)
    test_loss, test_acc = test(test_loader, model, criterion, device)
    train_loss_per_epoch.append(train_loss)
    train_acc_per_epoch.append(train_acc)
    test_loss_per_epoch.append(test_loss)
    test_acc_per_epoch.append(test_acc)
    print(
        f"Epoch {epoch + 1}/20: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")


## AlexNet
class AlexNet(nn.Module):
    def __init__(self):
        super(AlexNet, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=96, kernel_size=11, stride=4, padding=0)
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(kernel_size=3, stride=2)

        self.conv2 = nn.Conv2d(in_channels=96, out_channels=256, kernel_size=5, stride=1, padding=2)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(kernel_size=3, stride=2)

        self.conv3 = nn.Conv2d(in_channels=256, out_channels=384, kernel_size=3, stride=1, padding=1)
        self.relu3 = nn.ReLU()
        self.conv4 = nn.Conv2d(in_channels=384, out_channels=384, kernel_size=3, stride=1, padding=1)
        self.relu4 = nn.ReLU()
        self.conv5 = nn.Conv2d(in_channels=384, out_channels=256, kernel_size=3, stride=1, padding=1)
        self.relu5 = nn.ReLU()
        self.pool3 = nn.MaxPool2d(kernel_size=3, stride=2)

        self.fc1 = nn.Linear(256 * 6 * 6, 4096)
        self.relu6 = nn.ReLU()
        self.drop1 = nn.Dropout(p=0.5)
        self.fc2 = nn.Linear(4096, 4096)
        self.relu7 = nn.ReLU()
        self.drop2 = nn.Dropout(p=0.5)
        self.fc3 = nn.Linear(4096, 1000)

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu1(x)
        x = self.pool1(x)

        x = self.conv2(x)
        x = self.relu2(x)
        x = self.pool2(x)

        x = self.conv3(x)
        x = self.relu3(x)
        x = self.conv4(x)
        x = self.relu4(x)
        x = self.conv5(x)
        x = self.relu5(x)
        x = self.pool3(x)
        x = x.view(-1, 256 * 6 * 6)

        x = self.fc1(x)
        x = self.relu6(x)
        x = self.drop1(x)
        x = self.fc2(x)
        x = self.relu7(x)
        x = self.drop2(x)
        x = self.fc3(x)

        return x


# 2025-04-20
import torch
import torch.nn as nn
import torchvision
from torchsummary import summary

vgg = torchvision.models.vgg16()
resnet152 = torchvision.models.resnet152()
summary(vgg, (3, 224, 224))
#
summary(resnet152, (3, 224, 224))
```
