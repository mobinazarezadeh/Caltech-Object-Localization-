```
from sklearn.datasets import fetch_openml, fetch_california_housing, load_iris, load_digits
import numpy as np

print("در حال دانلود دیتاست‌ها... لطفاً شکیبا باشید (ممکن است چند دقیقه طول بکشد)")

# 1. دانلود دیتاست مسکن کالیفرنیا (California Housing)
print("1/3 Downloading California Housing...")
fetch_california_housing()

# 2. دانلود دیتاست اعداد دست‌نویس کوچک (Digits - نسخه سبک)
print("2/3 Loading Digits dataset...")
load_digits()

# 3. دانلود دیتاست معروف MNIST (اعداد دست‌نویس نسخه اصلی - حجیم حدود 50 مگابایت)
# این دیتاست برای دیپ لرنینگ عالی است
print("3/3 Downloading MNIST (this may take a while)...")
fetch_openml('mnist_784', version=1, as_frame=False)

# 4. دانلود دیتاست گل‌های زنبق (Iris - خیلی سبک)
load_iris()

print("\nتکمیل شد! ✅")
print("تمام این دیتاست‌ها در پوشه scikit_learn_data در سیستم شما ذخیره شدند.")
print("حالا حتی بدون اینترنت هم دستورات fetch کار خواهند کرد.")
```
