```
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
import torchvision
import matplotlib.pyplot as plt

to_tensor = torchvision.transforms.transforms.ToTensor()
train_dataset = torchvision.datasets.MNIST(root='./data/', train=True, download=True, transform=to_tensor)
test_dataset = torchvision.datasets.MNIST(root='./data/', train=False, download=True, transform=to_tensor)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)


def train(loader, model, criterion, optimizer):
    loss_per_epoch = []
    acc_per_epoch = []
    for data, label in loader:
        optimizer.zero_grad()
        data = data.view(-1, 1 * 28 * 28)  # Flattening image
        pred = model(data)
        loss = criterion(pred, label)
        loss.backward()
        optimizer.step()

        acc = accuracy_score(label, pred.argmax(dim=1))
        acc_per_epoch.append(acc)
        loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


def test(loader, model, criterion):
    loss_per_epoch = []
    acc_per_epoch = []
    with torch.no_grad():
        for data, label in loader:
            data = data.view(-1, 1 * 28 * 28)  # Flattening image
            pred = model(data)
            loss = criterion(pred, label)
            acc = accuracy_score(label, pred.argmax(dim=1))
            acc_per_epoch.append(acc)
            loss_per_epoch.append(loss.item())

    return torch.mean(torch.tensor(loss_per_epoch)), torch.mean(torch.tensor(acc_per_epoch))


class mnist_classifier(nn.Module):
    def __init__(self):
        super(mnist_classifier, self).__init__()
        self.fc1 = nn.Linear(784, 16)
        self.fc2 = nn.Linear(16, 16)
        self.fc3 = nn.Linear(16, 10)

        self.relu1 = nn.ReLU()
        self.relu2 = nn.ReLU()

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu1(x)
        x = self.fc2(x)
        x = self.relu2(x)
        x = self.fc3(x)
        return x


model = mnist_classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")


class mnist_classifier(nn.Module):
    def __init__(self):
        super(mnist_classifier, self).__init__()
        self.fc1 = nn.Linear(784, 100)
        self.fc2 = nn.Linear(100, 10)

        self.relu1 = nn.ReLU()

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu1(x)
        x = self.fc2(x)
        return x


model = mnist_classifier()
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters())  # w_new = w_old - lr * DL(w_old)

for epoch in range(10):
    train_loss, train_acc = train(train_loader, model, criterion, optimizer)
    test_loss, test_acc = test(test_loader, model, criterion)

    print(
        f"Epoch {epoch}/10: Train loss: {train_loss}, Train acc: {train_acc}, Test loss: {test_loss}, Test acc: {test_acc}")

test_img, test_label = next(iter(test_loader))

plt.imshow(test_img[1].squeeze(), cmap='gray')
test_label[7]
model(test_img[7].view(-1, 1 * 28 * 28))
torch.nn.Softmax()(model(test_img[7].view(-1, 1 * 28 * 28)))
torch.nn.Softmax()(model(test_img[1].view(-1, 1 * 28 * 28)))


```
